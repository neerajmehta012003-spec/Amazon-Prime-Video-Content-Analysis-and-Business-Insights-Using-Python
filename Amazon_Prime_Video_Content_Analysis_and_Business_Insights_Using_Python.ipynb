## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Create a copy of the merged dataset
wrangled_df = merged_df.copy()

# Check duplicate rows
print("Duplicate Rows Before:", wrangled_df.duplicated().sum())

# Remove duplicates
wrangled_df.drop_duplicates(inplace=True)

# Verify
print("Duplicate Rows After:", wrangled_df.duplicated().sum())

# Handling missing values
wrangled_df['description'] = wrangled_df['description'].fillna('Description Not Available')
wrangled_df['age_certification'] = wrangled_df['age_certification'].fillna('Not Rated')
wrangled_df['runtime'] = wrangled_df['runtime'].fillna(wrangled_df['runtime'].mean())
wrangled_df['imdb_score'] = wrangled_df['imdb_score'].fillna(wrangled_df['imdb_score'].mean())
wrangled_df['imdb_votes'] = wrangled_df['imdb_votes'].fillna(wrangled_df['imdb_votes'].mean())
wrangled_df['tmdb_popularity'] = wrangled_df['tmdb_popularity'].fillna(
    wrangled_df['tmdb_popularity'].median()
)
wrangled_df['tmdb_score'] = wrangled_df['tmdb_score'].fillna(
    wrangled_df['tmdb_score'].median()
)
wrangled_df['character'] = wrangled_df['character'].fillna('Not Applicable')

#Do not fill all missing values with 0. Only movies should have 0 seasons.
wrangled_df.loc[
    wrangled_df['type'] == 'MOVIE',
    'seasons'
] = wrangled_df.loc[
    wrangled_df['type'] == 'MOVIE',
    'seasons'
].fillna(0)

#Fill missing seasons for TV shows with the median number of seasons among TV shows.
show_median = wrangled_df.loc[
    wrangled_df['type'] == 'SHOW',
    'seasons'
].median()

wrangled_df.loc[
    wrangled_df['type'] == 'SHOW',
    'seasons'
] = wrangled_df.loc[
    wrangled_df['type'] == 'SHOW',
    'seasons'
].fillna(show_median)

# Remove extra spaces
text_columns = [
    'title',
    'description',
    'genres',
    'production_countries',
    'name',
    'character',
    'role'
]

for col in text_columns:
    wrangled_df[col] = wrangled_df[col].astype(str).str.strip()
````

### What all manipulations have you done and insights you found?

**Manipulations Performed*

* Merged titles_df and credits_df using id.
* Removed duplicate records.
* Handled missing values using appropriate strategies.
* Cleaned text fields and validated data quality.
* Converted multi-valued columns (genres, production_countries) into list format.
* Removed invalid records and verified the cleaned dataset.

**Insights**

* Final dataset contains 124,347 rows and 19 columns.
* Only 168 duplicate rows (0.14%) were found.
* Most missing values were concentrated in a small number of columns.
* The dataset covers Amazon Prime titles from 1912–2022.
